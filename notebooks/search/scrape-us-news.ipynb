```python
import pandas as pd
import os
import requests
from bs4 import BeautifulSoup
import re
import time
```

```python
from selenium import webdriver
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.common.by import By
```

```python
driver = webdriver.Firefox()
```

```python
url = 'https://www.usnews.com/best-colleges/rankings/national-universities'
```

```python
driver.get(url)
```

```python
results = []
```

```python
while True:
    names = driver.find_elements(By.CSS_SELECTOR, '.card-name')
    ranks = driver.find_elements(By.CSS_SELECTOR, '.ranked')

    result = pd.DataFrame({
        'name': [name.text for name in names],
        'rank': [rank.text for rank in ranks]
    })
    results.append(result)
    time.sleep(0.5)
```

```python
df = pd.concat(results)
```

```python
driver.close()
```
