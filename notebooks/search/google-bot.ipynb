```python
import requests
import pandas as pd
import json
import time
from dotenv import load_dotenv
import os
from ddgs import DDGS
from datetime import datetime
```

```python
load_dotenv('../google.env')
```

```python
def is_cds_title(title):
    title = title.lower()
    return 'cds' in title or 'common data set' in title or 'common dataset' in title or 'institutional research' in title
```

```python
def search_for_cds_files(unitid, domain):
    query = f"site:{domain} common data set"
    try:
        results = DDGS().text(query, max_results=1)
        print(f'Success on {domain}')
    except:
        print(f'No results found for {domain}')
        return None
    
    df = pd.DataFrame(results)
    df['unitid'] = unitid
    df['domain'] = domain

    return df
```

```python
directory = pd.read_csv('../../data/pgrm-control/search-space_w-landings.csv')
```

```python
directory = directory.fillna('')
directory = directory.query('file == "" and landing_page == ""').reset_index(drop=True)
```

```python
results = []
for i in range(0, len(directory['UNITID'])):
    unitid = directory['UNITID'][i]
    url = directory['WEBADDR'][i]
    results.append(search_for_cds_files(unitid, url))

    if i % 100 == 0:
        print(f'{i / len(directory["UNITID"])*100}% complete')
results = pd.concat(results)
```

```python
found_colleges = len(results['unitid'].unique())
success_rate = found_colleges / len(directory)
print(f'Success rate: {success_rate}')
```
