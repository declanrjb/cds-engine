```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import os
```

```python
df = pd.read_csv('../../data/pgrm-control/search-space.csv')
```

```python
def extract_peers(unitid):
    if type(unitid) is not str:
        unitid = str(unitid)
    url = f'https://nces.ed.gov/IPEDS/DFR/2025/ReportHTML.aspx?unitid={unitid}'

    response = requests.get(url)
    if response.status_code != 200:
        raise ValueError('URL not found')
    
    soup = BeautifulSoup(response.text)

    anchors = soup.select('.compgroup a')
    anchors = [anchor for anchor in anchors if 'unitid' in anchor['href'].lower()]

    df = pd.DataFrame({
        'peer_name': [anchor.text for anchor in anchors],
        'peer_unitid': [anchor['href'].split('=')[1] for anchor in anchors]
    })
    df['selecting_college'] = unitid

    return df
```

```python
for i in range(0, len(df['UNITID'])):
    unitid = df['UNITID'][i]
    outpath = f'../../data/peers/{unitid}.csv'
    if not os.path.isfile(outpath):
        temp = extract_peers(unitid)
        temp.to_csv(outpath, index=False)
        print(f'Succeeded on {unitid}...')
        time.sleep(1)
    else:
        print(f'{unitid} already exists')
    
    if i % 100 == 0:
        print(f"{i / len(df['UNITID']) * 100}% complete")
```
